# Heat Resilience in Sight — Code 2: Hourly UTCI climatology

Download 2015–2024 **hourly** ERA5-land fields at each city centroid (Open-Meteo archive). For 10:00–17:00 local time, compute UTCI with simultaneous air temperature, humidity, and wind.

- Open-sun $T_\mathrm{mrt}=T_\mathrm{air}+\Delta T_\mathrm{mrt}$ from SolarCal (standardised standing pedestrian).
- Shaded bound $T_\mathrm{mrt}=T_\mathrm{air}$.
- Daily UTCI is the daytime maximum; demand is $\mathrm{mean}\,\max(U-26,0)$.

**Cache:** `data/climate_cache/hourly_v1/{city}.json` (old daily JSON is never reused).

```python
import sys
from pathlib import Path

ROOT = Path('.').resolve()
sys.path.insert(0, str(ROOT))

from scripts.climate import CACHE_VERSION, panel_climatology
from scripts.config import CACHE_HOURLY, CITIES, PANEL_ORDER, U0

print('cache:', CACHE_HOURLY, 'version', CACHE_VERSION)
print('U0 (moderate heat):', U0)
```

## 1. Monthly open-sun and shaded demand

Warming columns add a uniform $+1.5$/$+2.5\,^\circ$C to $T_\mathrm{db}$ and $T_\mathrm{mrt}$ before recomputing UTCI.

```python
monthly = panel_climatology()
annual = monthly.groupby('city')[['utci', 'demand', 'utci_shade', 'demand_shade']].mean()
annual['label'] = [CITIES[k]['label'] for k in annual.index]
annual['koppen'] = [CITIES[k]['koppen'] for k in annual.index]
annual.loc[PANEL_ORDER, ['label', 'koppen', 'utci', 'demand', 'demand_shade']].round(3)
```

```python
hk_jja = monthly[(monthly.city == 'svi_hk') & monthly.month.isin([6, 7, 8])]
print('Hong Kong JJA open-sun demand:', round(float(hk_jja.demand.mean()), 3))
print('Hong Kong JJA shaded demand:', round(float(hk_jja.demand_shade.mean()), 3))
print('Hong Kong annual open-sun demand:', round(float(monthly[monthly.city == 'svi_hk'].demand.mean()), 3))
```
