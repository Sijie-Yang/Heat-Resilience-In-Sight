# Heat Resilience in Sight — Code 4: VATA and SHR maps

Regular 800 m hexes in a local metre frame (not matplotlib `hexbin`).

- VATA maps / distributions: Hong Kong mean as the supply anchor.
- SHR maps: $\mathrm{SHR}_i = V_i - V^{\mathrm{need}}_{c,t}$. Peak-heat quarter in the main figure; DJF/MAM/JJA/SON in the appendix.

No 92k Singapore set.

```python
import sys
from pathlib import Path

import pandas as pd

ROOT = Path('.').resolve()
sys.path.insert(0, str(ROOT))

from scripts.config import CALIBRATE_KEY, FIGS, PANEL_ORDER, TABLES
from scripts.maps import load_panel_points, write_shr_spatial_figures, write_vata_spatial_figures
from scripts.panel import summarise_panel

stats = summarise_panel(write=True)
monthly = pd.read_csv(TABLES / 'monthly_climate_vata.csv')
print('baseline', CALIBRATE_KEY, round(float(stats.loc[CALIBRATE_KEY, 'mean']), 3))
print(stats.loc[PANEL_ORDER, ['label', 'koppen', 'n', 'mean']].round(3))
```

```python
panel = load_panel_points()
write_vata_spatial_figures(stats, FIGS, panel=panel)
write_shr_spatial_figures(monthly, FIGS, panel=panel)
for name in [
    'fig_vata_distributions.png',
    'fig_vata_maps.png',
    'fig_shr_maps.png',
    'fig_shr_quarterly_a.png',
    'fig_shr_quarterly_b.png',
]:
    print('wrote', FIGS / name)
```
